```python
#min/max , sorted 복습
nums=[1,-3,5,4]
print(max(nums))
print(min(nums))

names = ["banana","grape","apple"]
print(max(names))  #사전순서상 가장 나중에
print(min(names))  #사전순서상 가장 먼저

nums=[1,-3,5,4]
print(sorted(nums))  #sorted()함수는 오름차순으로 정렬된 리스트를 반환해줌

names=["banana","grape","apple"]
print(sorted(names))  #sorted()함수는 오름차순 정렬
```

```python
#key입력사용 / lambda
#max(),min(),sorted()에  key를 추가하여 원하는 기준으로 정렬할수잇다
#형식은
#max(입력,key=함수)
#max(입력,key=lambda x: x.내장함수())
nums = [1,-3,4,5]
print(sorted(nums,key=abs)) #nums를 정렬하라 key=abs,절대값 크기를 기준으로

names=["apple","kiwi","banana"]
print(sorted(names,key=len))  #names를 정렬하라, key=len 문자열 길이를 기준으로

names=["BANANA","GRAPE","apple"]
print(sorted(names))
print(sorted(names,key=lambda x: x.lower())) #names를 정렬하라, key=lambda를 기준으로, lambda가 무엇인가.오름차순으로

nums=[[5,6],[-30,-10,-10]]
print(max(nums, key=len))
print(max(nums,key=sum))
```

```python
#zip()함수
#둘 이상의 반복가능한 값을 입력받아 인덱스번호같은것끼리 짝꿍, 튜플형태로 짝꿍 반환
nums=[1,2,3]
names=["one","two","three"]
for item in zip(nums,names):
  print(item)

#세개이상의 입력,#인덱스순서대로 하나씩 꺼내서 출력함
nums=[1,2,3]
names=["one","two","three"]
num_strs="123"
for item in zip(nums,names,num_strs):
  print(item)

#길이가 다른 경우
#길이가 짧은 쪽에 맞춰지고 긴쪽은 짝이 없어서 버려짐 ㅜㅜ
nums=[1,2,3]  #3이 짝이 없다 ㅠㅠ
names=["one","two"]
for item in zip(nums,names):
  print(item)
#튜플형태 말고 다른 형태로도 ㄱㄴ
nums=[1,2,3]
names=["one","two"]
for i,j in zip(nums,names):
  print("num: {} / names:{}".format(i,j))

names=["Kim","Lee"]
ages=[20,30]
for n, a in zip(names,ages):
  print(n,a)
```

```python
#enumerate()함수 열거함수,인덱스값과 요소를 반환함
names=["apple","grape","banana"]
for i, name in enumerate(names):
  print("{}:{}".format(i,name))

names=["apple","grape","banana"]
for i, names in enumerate(names,1):
  print("{}: {}".format(i,names))

chars=["*","+","#","&"]
for i,c in enumerate(chars):
  print(c*i)

chars=["*","+","#","&"]
for i,c in enumerate(chars,1):
  print(c*i)
```

```python
#isinstance()함수 , 값의 타입(자료형)을 알고싶을때 사용함
#isinstance(값,타입)

if isinstance("python",str):  #str->string형 문자열
  print("python is string")

num= 41
if isinstance(num,int):
  print("{} is int".format(num)) #41은 정수,int형
if isinstance(num,float):
  print("{}is float".format()) #41.0 이float형

#type이 여러개인 경우
#isinstance(rkqt,(타입1,타입2,타입3,...))

if isinstance([1,2],(list,tuple)):
  print("[1,2] is list or tuple")

num=41
if isinstance(num,(int,float)):
  print("{} is number".format(num))
if isinstance(num,(list,tuple)):  #false
  print("{} is list or tuple".format(num))

args=(3,4,5)
if isinstance(args,tuple):
  print(sum(args))
elif isinstance(args,int):  #false
  print(args)

args=3.5
if isinstance(args, tuple):  #false
  print(sum(args))
elif isinstance(args,(int,float)):
  print(args)
```

```python
#all,any함수
#all()함수-> 다 True여야함, 하나라도 False이면 False를 출력 /and에 대응됨
#any()함수-> 하나라도 True이면 True를 출력 / or()에 대응됨

num = 30
check_num=[num%2==0,num%3==0,num%5==0]
if all(check_num):
  print("all:and")

num=12
check_num=[num%2==0,num%3==0,num%5==0]
if all(check_num):
  print("all:and")

num = 8
check_num=[num%2==0,num%3==0,num%5==0]
if any(check_num):
  print("any:or")

print(all([True,True,False,True])) #all과 any도함수라 묶어줘야함..
print(all([True,True,True,True]))
print(any([False,False,True]))
print(any([False,False]))
```

```python
#문자열 format확장,format()함수는 중괄호 안에 들어감
num = 3
name="apple"
print("num={},name={}".format(num,name))

#소수점 자리 수
#무한소수,...문자열 길이가 길어짐 -> {:.f}옵션을 사용하여 특정 자리수까지만 문자열로 나타내게 된다
#"{:.자리수f}".format(float변수)
print("{:.1f}".format(0.6666))

num = 1/3
print("num={}".format(num))
print("num={:.3f}".format(num))

#숫자 길이{:x}, x자수로 출력되게 함
#문자열의 길이를 원하는 대로 맞출수잇음
print("{:10}".format("python")) #문자열은 왼쪽정렬, 숫자는 오른쪽정렬

#숫자는 오른쪽 정렬
num=41
print("num={}".format(num))
print("num={:5}".format(num))

for i in range(11): #0-10
  print("{:02}".format(i))  #00,01,02...

num = 41.123
print("num={:7.2f}".format(num))  #7자리로,소수점 둘째자리까지 출력

#1000자리마다 콤마 {:,} 붙여주면 됨
num=1234567
print("num={}".format(num))
print("num{:,}".format(num))

print(1/7)
print("{:.3f}".format(1/7))  #소수점 세자리까지 출력
print("{:3}".format(7)) #세칸(?)으로 출력함
print("{:03}".format(7))  #세칸으로 출력하는데 공백칸은 0으로 채움

c=299792458
print("speed: {}".format(c))
print("speed: {:,}".format(c))
```

```python
#리스트 컴프리헨션(list comprehension)->리스트,딕셔너리,집합에 사용
# C for A in B
#규칙적인 값으로 이루어진 리스트를 생성하는 방법

#1
num_strings = []
for i in range(5):
  num_strings.append(str(i))
print(num_strings)
#1을 list comprehension으로
num_strings = [str(i) for i in range(5)]
print(num_strings)

#for if 리스트 컴프리헨션
#1
two_multiples=[]
for i in range(10):
  if i%2 ==0:
    two_multiples.append(str(i))
print(two_multiples)

#1을 list comprhension으로
two_multiples=[str(i) for i in range(10) if i%2==0]

#다중 for문 리스트 컴프리헨션,for문은 들여쓰기 단위로 생각하기
#1
animals=[]
for i in ["slow","fast"]:
  for j in ["dog","cat"]:
    animals.append("{} {}".format(i,j))
print(animals)
#1 list comprehension 걍 쭉쓰면 됨,,? 상위는 첨에 등장한것,하위는 나중에 등장한 것
animals=[{i,j} for i in ["slow","fast"] for j in ["dog","cat"]]
print(animals)

squares=[i**2 for i in range(5)]
print(squares)

odd_squares=[i**2 for i in range(5) if i % 2 ==1]
print(odd_squares)
```
