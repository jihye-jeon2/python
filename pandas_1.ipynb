-iloc는 정수 기반 인덱싱을 사용, 행이나 열을 위치(순서)로 , 즉 인덱스 번호로 선택할때 사용

```python
#행과 열을 함께 지정하여 추출하기
display(df.loc[42,'country']) #loc[행이름,열이름]
```

```python
#여러 행과 열 지정하여 데이터 추출하기
display(df.loc[[0,99,999],['country','lifeExp','gdpPercap']])  #행이름0,9,99,999  숫자임 열이름 여러개니까
```

```python
display(df.iloc([0,99,999],[0,3,5]))   #iloc는 인덱스번호로 , loc는 이름으로
```

```python
display(df.groupby('year')['lifeExp'].mean())
```

```python
#2개 이상 열 그룹화하기
multi_group_var=df.groupby(['year','continent'])[['lifeExp','gdpPercap']].mean()
display(multi_group_var)
```

```python
flat=multi_group_var.reset_index() #year,continent 일반 열로 바뀌고, 새로운 0부터 시작하는 인덱스를 생
display(flat)
```

```python
#그룹화한 데이터개수 구하기
print(df.groupby('continent')['country'].nunique()) #고유한(unique)값들의 개수를 반환
```

```python
print(df.groupby('continent')['country'].value_counts())
#선택된 'country'열에서 각 대륙별로 각 국가가 몇번나타나는지를 계산
#Series 형태로 반환하고 가장 빈도가 높은 값부터 내림차순으로 정렬
#Series 내의 고유한 값들과 각 값의 출현 빈도를 계산
```
