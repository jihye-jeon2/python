```python
#if-elif 문
#구조는
#if 조건식1:
  #실행코드1
#elif 조건식2:
  #실행코드2
#elif 조건식3
  #실행코드3

#ex_1
if 82 > 90:
  print("A grade")
elif 82>80:
  print("B grade")  # True이면 여기서 끝냄
elif 82>70:
  print("C grade")

#ex_2
score = 68
if score >= 70:
  print("A")
elif score >= 60: #B실행되고 끝남
  print("B")
elif score >= 50:
  print("C")

score = 85
if score > 90:
  print("A grade")
elif score>80:
  print("B grade")
elif score>70:
  print("C grade")
```
