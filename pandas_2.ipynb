###Series시리즈와 DataFrame데이터프레임

```python
import pandas as pd
```

```python
#시리즈 만들기  리스트형태로 넣어주면 됨
s=pd.Series(['apple','banana']) #파이썬 리스트(list)를 pd.Series()함수에 전달하면 리스트 각요소가 시리즈의 데이터가 된
display(s)
```

```python
s=pd.Series(data=['Wes Mckinney', 'Creator of Pandas'], index=['Person','Who'])   #인덱스를 숫자가 아닌 이름으로 지정 ㄱ
display(s)
```

```python
#데이터프레임 만들기
scientists =pd.DataFrame(
    data={
    "Occupation":["Chemist","Statistician"],
    "Born":["1920-07-25","1876-06-13"],
    "Died":["1958-04-16","1937-10-16"],
    "Age":[37,61]

},
    index=["Rosaline Franklin","William Gosset"],
    columns=["Occupation","Born","Died","Age"]
)
display(scientists)
```

```python
print(first_row.index)
```

###열 내용 변경하기

```python
print(scientists['Age'])
```

```python
print(scientists["Age"].sample(frac=1,random_state=42))
#frac=1 frac 인자는 샘플링할 비율을 지정, 전체데이터(100%)를 샘플링하겟다는 의미,
#모든 데이터를 포함하되 순서만 무작위로 섞는 (셔플링) 효과를 줍니다
#random_state인자는 난수생성기의 시드(seed)를 설정,
# 이 값을 지정하면 코드를 여러번 실행해도 항상 동일한 순서로 데이터가 섞이게된다.(=결과의 재현성을 보장한다)
```

```python
scientists["Age"]=scientists["Age"].sample(frac=1,random_state=42)  #섞기 전이랑 동일함
print(scientists['Age'])
```

```python
from typing import ValuesView
scientists["Age"] = scientists["Age"].sample(frac=1,random_state=42).values
print(scientists['Age'])
```
