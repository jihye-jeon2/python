## 1. 객체와 클래스의 이해
* **객체 지향 프로그래밍(Object Oriented Programming)**: 현실 세계의 모든 사물을 '객체'로 보고, 이를 프로그램으로 구현하는 기법입니다.
* **객체(Object)**: 현실 세계에 존재하는 모든 것이 객체가 될 수 있으며, **속성**과 **메서드**로 구성됩니다.
    * **속성(Attribute)**: 객체의 상태나 특징을 나타내는 변수입니다.
    * **메서드(Method)**: 객체의 동작이나 기능을 나타내는 함수입니다.
* **클래스(Class)**: 객체를 만들기 위한 '설계도' 또는 '틀'입니다. 클래스를 통해 생성된 실제 사물을 '인스턴스(Instance)' 또는 '객체'라고 부릅니다.

```python
class Account:
  def  __init__(self):  #define
    self.balance = 0
  def deposit(self,amount):
    self.balance +=amount
    return self.balance

  def withdraw(self, amount):
    self.balance -=amount
    return self.balance

#a1 = Account()  # Account라는 클래스의 인스턴스 또는 객체
#a2 = Account()

#print(a1.deposit(100))  #입금
#print(a1.withdraw(50))   #출금

#print(a2.deposit(200))
#print(a2.withdraw(100))
```

## 2. 클래스 생성 및 사용
* **클래스 정의**: `class` 키워드를 사용하며, 내부에 속성(변수)과 메서드(함수)를 정의합니다.
* **생성자 (`__init__`)**: 객체가 생성될 때 자동으로 호출되는 특별한 메서드로, 주로 객체의 초기 상태(속성)를 설정하는 데 사용됩니다.
* **Self 키워드**: 메서드 내부에서 객체 자신의 속성에 접근할 때 사용하며, 모든 메서드의 첫 번째 매개변수로 지정해야 합니다.
* **정보 은닉 (Private)**: 속성 이름 앞에 `__`(언더바 두 개)를 붙여 클래스 외부에서 직접 수정하지 못하도록 보호할 수 있습니다.

## 3. 상속 (Inheritance)
* **개념**: 기존 클래스(부모 클래스)의 속성과 메서드를 그대로 물려받아 새로운 클래스(자식 클래스)를 만드는 기능입니다.
* **장점**: 코드의 재사용성을 높여주며, 중복된 코드를 줄여 유지보수가 쉬워집니다.
* **메서드 오버라이딩(Overriding)**: 부모 클래스로부터 물려받은 메서드를 자식 클래스에서 용도에 맞게 새롭게 재정의하여 사용하는 것을 말합니다.

```python
#부모 클래스
class Student:
  def __init__(self,name,computer):
    self.name = name
    self.computer = computer

  def set_computer(self,computer):
    self.computer = computer

  def get_name(self):
    return self.name
```

```python
# 이과에 해당하는 자식 클래스
#class 자식클래스이름(부모클래스이름):
class Science(Student):
  def __init__(self,name,computer,science):  # 'math' 인자를 제거하여 호출부와 일치시킴
    super().__init__(name,computer)  #부모클래스의 생성자 호출
    self.science = science

  def set_science(self,science):
    self.science = science # 오타 수정: self.scince -> self.science


  def get_average(self):
    return(self.science+self.computer)/2
```

```python
#문과에 해당하는 자식클래스
class Social(Student):
  def __init__(self,name,computer,social):
    super().__init__(name,computer)
    self.social = social


  def set_social(self,social):
    self.social = social

  def get_average(self):
    return(self.social + self.computer) / 2
```

```python
#자식클래스를 통한 객체를 만들기
st1 = Science('lee',90,80)   # student1에 해당하는 객체 생성 self->st1,name->lee computer->90,science->80
print(st1.get_name())
print(st1.get_average())
st1.set_computer(100)

print(st1.get_average())
```

```python
st2 = Social('kim',96,88)
print(st2.get_name())
print(st2.get_average())
st2.set_social(96)
print(st2.get_average())
```

```python
#클래스 Circle 만들기
# 반지름 속성 __radius
#반지름 속성을 확인하는 메서드 get_radius
# 반지름 설정하는 set_radius 메서드
#원의 넓이 구하는 get_area메서드
#원의 둘레를 구하는 get_circumference 메서드

import math

class Circle:
  def __init__(self, radius = 10):#생성자 함수(객체가 생성되엇을때 자동으로 생성되는 함수)
    self.__radius = radius

  def get_radius(self): #get은 가져올때 set은 지정해줄때 사용하는 함수라함
   return self.__radius

  def set_radius(self, radius):
    self.__radius = radius

  def get_area(self):
    return math.pi*self.__radius**2

  def get_circumference(self):
    return 2*math.pi*self.__radius

c1= Circle(7)
print(f'반지름: {c1.get_radius()}')
c1.set_radius(15)
print(f'반지름 : {c1.get_radius()}')
print(f'넓이: {c1.get_area():.2f}')
print(f'둘레 : {c1.get_circumference():.2f}')
```

### 🚗 자동차 클래스와 객체
1. **클래스 설계**: `Car` 클래스를 생성하고, 모델명(`__model`)과 연비(`__fuelefficiency`)를 속성으로 정의합니다.
2. **정보 보호**: 속성들을 `private`으로 설정하여 외부의 직접 접근을 막고, 안전하게 값을 읽거나 수정할 수 있는 전용 메서드(`get_model`, `set_fuelefficiency` 등)를 만듭니다.
3. **연료량 계산 로직**: 주행 거리를 입력받아 (주행 거리 / 연비) 공식을 통해 필요한 연료량을 계산하여 반환하는 메서드를 구현합니다.
4. **객체 생성**: 정의된 클래스를 바탕으로 `car1`, `car2` 등 서로 다른 모델명과 연비를 가진 실제 자동차 객체들을 만들어 실행합니다.

```python
# 'GV70' . 12
#distance = input(주행거리: )

class Car:
  def __init__(self,model,fuelefficiency):
      self.__model = model
      self.__fuelefficiency = fuelefficiency

  def get_model(self):
    return self.__model

  def get_fuelefficiency(self):
    return self.__fuelefficiency

  def set_fueldfficiency(self,fuelefficiency):
    self.__fuelefficiency = fuelefficiency

  def get_fuelammount(self,distance):
    return distance/self.__fuelefficiency





```

```python
car1 = Car('GV70',12)
print(f'모델명: {car1.get_model()}')
print(f'연비: {car1.get_fuelefficiency()}')
distance = float(input('주행거리: '))
print(f'연료량:{car1.get_fuelammount(distance)}')




```
