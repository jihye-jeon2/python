```python
#필요한 라이브러리 임포트
import pandas as pd  #pd를 별명으로사용하겟다
import numpy as np
from sklearn.datasets import load_iris


```

```python
#iris 데이터셋 불러오기
iris = load_iris()
print(iris)


```

pandas dataFrame

```python
#데이터프레임으로 변환
df = pd.DataFrame(iris.data, columns=['sepal_length','sepal_width','petal_length','petal_width'])
df['species'] = pd.Categorical.from_codes(iris.target,iris.target_names)
```

```python
print('\n=== 결측치 확인 ===')  #결측치 -> 비어잇는 값
print(df.isnull().sum())
```

```python
print('\n=== 종류별 데이터 개수 ===')
print(df['species'].value_counts())
```

평균 (Mean)

```python
#단일 칼럼 평균
print('꽃받침 길이 평균: ',df['sepal_length'].mean())
print(f'꽃받침 길이 평균: {df['sepal_length'].mean():.2f} ')
```

```python
#전체 숫자형 칼럼 평균
print('\n전체 칼럼 평균:')
print(df.mean(numeric_only=True).round(2))
#(numeric_only ->True),숫자형칼럼(정수,실수)만 계산
```

```python
#종류별 평균 비교
print('\n종류별 꽃입 길이 평균: ')
print(df.groupby('species')['petal_length'].mean())
```

중앙값(Median)

```python
#단일 칼럼 중앙값
print('꽃받침 길이 중앙값:',df['sepal_length'].median())
```

```python
#전체 칼럼 중앙값
print('\n전체 칼럼 중앙값:')
print(df.median(numeric_only=True))  #숫자형에 해당하는 애들만 중앙값 ㄱㄱ
```

```python
#평균 vs중앙값 비교  -> 데이터의 분포도를 알수 0
print('\n=== 평균 vs 중앙값 비교 ===')
comparison = pd.DataFrame({
    '평균':df.mean(numeric_only=True).round(2),
    '중앙값': df.median(numeric_only=True)
})
print(comparison)
```

분위수(Quantile/Percentile)

```python
#분위수 구하기
print("=== 분위수(25%,50%,75%) ===")
print(df.quantile([0.25,0.5,0.75],numeric_only=True)) #0.5는 중앙값과 동일
```

```python
#특정 칼럼의 분위수
col='sepal_length'
Q1=df[col].quantile(0.25)
Q2=df[col].quantile(0.5)
Q3=df[col].quantile(0.75)
IQR = Q3-Q1
print(f"\n꽃받침 길이 분위수:")
print(f"  Q1 25%: {Q1}")
print(f" Q2 50%: {Q2}" )
print(f" Q3 75%: {Q3}" )
print(f" IQR     : {IQR} ")
```

```python
#이상치 경계값 벗어나는 값들은 이상한 값들
lower = Q1 - 2.5 * IQR
upper = Q3 +1.5 * IQR
print(f'\n이상치 경계값: ')
print(f' 하한: {lower:.2f}')
print(f' 상한: {upper:.2f}')
```

```python
# 이상치 확인
outliers = df[(df[col]<lower) | (df[col]>upper)]
print(f'\n이상치 데이터 수: {len(outliers)}개')
# df[True,False,True] DataFrame에 불리언 시리즈를 전달하면 True로 표시된 행들만 필터링하여 새로운 DataFrame인 outliers를 반환
# True로 표시된 행들만 필터링하여 새로운 DataFrame인 outliers를 반환
```

분산(Variance)

```python
#데이터가 넓은분포(크다) or 좁은 분포(작다)
#단일 칼럼 분산
print(f"꽃받침 길이 분산:{df['sepal_length'].var():.2f} ")
```

```python
#전체 칼럼 분산
print('\n===전체 칼럼분산 ===')
print(df.var(numeric_only=True))
```

```python
#종류별 분산 비교(분산이 크면 데이터가 넓게 퍼져 잇음)
print('\n ===종류별 길이 분산: ')
print(df.groupby('species')['petal_length'].var().round(2))
```

표준편차 (Standard Deciation)

```python
#단일 칼럼 표준편차
print("꽃받침 길이 표준편차: ", df['sepal_length'].std())
print(f'꽃받침 길이 표준편차: {df["sepal_length"].std():.2f}')
```

```python
#전체 칼럼 표준편차
print('\n === 전체 칼럼 표준편차 ===')
print(df.std(numeric_only=True).round(2))
```

```python
#분산 vs 표준편차 비교 정리
print('\n === 분산 vs 표준편차 비교 ===')
summary = pd.DataFrame({
    '분산':df.var(numeric_only=True).round(4),
    '표준편차':df.std(numeric_only=True).round(4)
})
print(summary)
#표준편차: 분산의 제곱근으로,
#데이터가 평균으로부터 평균적으로 얼마나 떨어져 잇는지를 나타냄
#데이터의 원래 단위를 따르기 때문에 분산보다 직관적으로 이해하기 쉬움

#표준편차 크다,데이터 포인트들이 서로크게 다르다는 것을의미
#표준편차 작다,데이터 포인트들이 서로 유사하다는 것을 의미
```

```python
# 평균 +- 표준편차 범위(데이터의 약 68%가 여기에 속함)
col='sepal_length'
mean = df[col].mean()
std = df[col].std()
print(f"\n 꽃받침 길이: 평균 +- 표준편차")
print(f"평균:{mean:.2f}")
print(f" 표준편차: {std:.2f}")
print(f"범위: {mean-std:.2f}-{mean + std:.2f}")
count_in_range = df[(df[col] >= mean-std) & (df[col] <= mean +std) ].shape[0]
#True 로 표시된 행들만 필터링하여 새로운 DataFrame인 count_in_range을 반환

#.shape[0]:필터링된 DataFrame의 행의 개수를 반환
print(f"  이 범위에 속하는 데이터: {count_in_range}개 ({count_in_range/len(df)*100:.1f}%)")
```

```python
df[(df[col] >= mean - std) & (df[col] <= mean + std)]
```

```python
#내가 원하는 통계만 선택해서보기
print('\n=== 커스텀 기술통계 ===')
customs_stats = df.groupby('species')['petal_Length'].agg([
    ('평균','mean'),
    ('중앙값','median'),
    ('표준편차','std'),
    ('최소값','min')
    ('최대값','max')
]).round(2)
print(customs_stats)
```

###넘파이(Numpy)의 기술통계함수

```python
import numpy as np
data=df['sepal_length'].values  #numpy배열로 변환->array어쩌구~~
```

```python
data
```

```python
print("===Numpy 기술통계 함수 ===")
print(f"np.mean() 평균: {np.mean(data):.4f}")
print(f"np.median() 중앙값: {np.median(data):.4f}")
print(f"np.min() 최솟값: {np.min(data):.4f}")
print(f"np.max() 최댓값: {np.max(data):.4f}")
print(f"np.var() 분산(모집단): {np.var(data):.4f}")
print(f"np.var (ddof=1) 분산(표본):{np.var(data,ddof=1):.4f}")
print(f"np.std() 표준편차(모집단): {np.std(data):.4f}")
print(f"np.std(ddof=1) 표준편차(표본):{np.std(data,ddof=1):.4f}")
print(f"np.percentile(25) Q1 1사분위: {np.percentile(data,25):.4f}")   #25% 50% 75% 100%
print(f"np.percentile(50) Q2 2사분위: {np.percentile(data,50):.4f}")
print(f"np.percentile(75) Q3 3사분위: {np.percentile(data,75):.4f}")
print(f"np.ptp() 범위(max-min): {np.ptp(data):.4f}")   #Range
```

```python
#다중 백분위수 한번에계산
print("\n=== 다중 백분위수===")
percentile=np.percentile(data,[10,25,50,75,90])#10% 25% 50% 75% 90% 백분위
percentile
```

```python
for p,v in zip([10,25,50,75,90].percentiles):
  print(f"{p}번쨰 백분위수:{v:.2f}")
```
