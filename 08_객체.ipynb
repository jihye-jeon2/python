## 1. 객체와 클래스의 이해
* **객체 지향 프로그래밍(Object Oriented Programming)**: 현실 세계의 모든 사물을 '객체'로 보고, 이를 프로그램으로 구현하는 기법입니다.
* **객체(Object)**: 현실 세계에 존재하는 모든 것이 객체가 될 수 있으며, **속성**과 **메서드**로 구성됩니다.
    * **속성(Attribute)**: 객체의 상태나 특징을 나타내는 변수입니다.
    * **메서드(Method)**: 객체의 동작이나 기능을 나타내는 함수입니다.
* **클래스(Class)**: 객체를 만들기 위한 '설계도' 또는 '틀'입니다. 클래스를 통해 생성된 실제 사물을 '인스턴스(Instance)' 또는 '객체'라고 부릅니다.

```python
class Account:
  def  __init__(self):  #define
    self.balance = 0
  def deposit(self,amount):
    self.balance +=amount
    return self.balance

  def withdraw(self, amount):
    self.balance -=amount
    return self.balance

#a1 = Account()  # Account라는 클래스의 인스턴스 또는 객체
#a2 = Account()

#print(a1.deposit(100))  #입금
#print(a1.withdraw(50))   #출금

#print(a2.deposit(200))
#print(a2.withdraw(100))
```

## 2. 클래스 생성 및 사용
* **클래스 정의**: `class` 키워드를 사용하며, 내부에 속성(변수)과 메서드(함수)를 정의합니다.
* **생성자 (`__init__`)**: 객체가 생성될 때 자동으로 호출되는 특별한 메서드로, 주로 객체의 초기 상태(속성)를 설정하는 데 사용됩니다.
* **Self 키워드**: 메서드 내부에서 객체 자신의 속성에 접근할 때 사용하며, 모든 메서드의 첫 번째 매개변수로 지정해야 합니다.
* **정보 은닉 (Private)**: 속성 이름 앞에 `__`(언더바 두 개)를 붙여 클래스 외부에서 직접 수정하지 못하도록 보호할 수 있습니다.

## 3. 상속 (Inheritance)
* **개념**: 기존 클래스(부모 클래스)의 속성과 메서드를 그대로 물려받아 새로운 클래스(자식 클래스)를 만드는 기능입니다.
* **장점**: 코드의 재사용성을 높여주며, 중복된 코드를 줄여 유지보수가 쉬워집니다.
* **메서드 오버라이딩(Overriding)**: 부모 클래스로부터 물려받은 메서드를 자식 클래스에서 용도에 맞게 새롭게 재정의하여 사용하는 것을 말합니다.

```python
#private  속성 값을 얻을려면 속성 값을 반환하는 메서드를 이용해야 됨
class Account:
  def  __init__(self, amount= 0):  #define
    self.__balance = amount #__balance 는 private속성

  def deposit(self,amount):
    self.__balance +=amount
    return self.__balance

  def withdraw(self, amount):
    self.__balance -=amount
    return self.__balance
  def get_balance(self):
    return self.__balance

  a1 = Account(50)
  print(a1,get_balance())

```

```python
#클래스 Circle 만들기
# 반지름 속성 __radius
#반지름 속성을 확인하는 메서드 get_radius
# 반지름 설정하는 set_radius 메서드
#원의 넓이 구하는 get_area메서드
#원의 둘레를 구하는 get_circumference 메서드

import math

class Circle:
  def __init__(self, radius = 10):#생성자 함수(객체가 생성되엇을때 자동으로 생성되는 함수)
    self.__radius = radius

  def get_radius(self): #get은 가져올때 set은 지정해줄때 사용하는 함수라함
   return self.__radius

  def set_radius(self, radius):
    self.__radius = radius

  def get_area(self):
    return math.pi*self.__radius**2

  def get_circumference(self):
    return 2*math.pi*self.__radius

c1= Circle(7)
print(f'반지름: {c1.get_radius()}')
c1.set_radius(15)
print(f'반지름 : {c1.get_radius()}')
print(f'넓이: {c1.get_area():.2f}')
print(f'둘레 : {c1.get_circumference():.2f}')
```
