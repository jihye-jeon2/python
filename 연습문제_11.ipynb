```python
# @title
#all(),any() 복습

TFs = [True,True,False]
print(all(TFs))  #다 True여야 True
print(any(TFs))  #하나라도 True면 True

print(all([True,True,True]))
print(any([False,False,False]))

#all(),any()응용 -> generator expression(?)
#list comprehension 이란 무엇인가 한줄 간단한 코드, 가독성 떨어질수잇지만 간단
#all(A for B in C)
#any(A for B in C)
all(i%2==0 for i in range(5))
any(i%2==0 for i in range(5))

nums=[2,4,5,6,8,10]
print(all(n%2==0 for n in nums))

files = ["a1.jpg","a2.jpg","a3.txt"]
print(any(f[-4:]==".txt" for f in files)) #f의 오른쪽에서 4~끝까지가 .txt인게 하나라도 잇으면 True출력

words=["ape","apple","alligator"]
print(all(w[0] == "a" for w in words))  #words리스트에 잇는 요소들을 w에다가
print(all(len(w)>5 for w in words))

```
