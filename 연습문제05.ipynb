```python
#멤버연산자
#뒤에 잇는 리스트(?)의 멤버인지 체킹할수잇는 연산자
print(1 in [1,2,3])
print(1 not in [1,2,3])
```

#논리 연산자
#논리합,논리곱,부정연산자
# a and b , a or b, not a

print(4>3 and 5<4) #false
print(4>3 or 5<4) #true

```python
num = 3
print(num<5)
print(num>4 and num<8)
print(num>4 or num < 8)
```

```python
#멤버연산자->뒤에 리스트(?)에 잇는지 확인함
num = 3
print(num in [3,6,9])
print(num in ["3","6","9"])  # 숫자형과 문자형"3"
```

IF문,조건문 실행해야 할 코드에 들여쓰기해주는게 중요함(조건문에서 제어되는코드는 들여쓰기로 구분함)

```python
#if 조건식:
  #실행코드1(조건식이 참을경우 실행되는 코드)
#실행코드2(if문 빠져나옴)

if sum([1,2.3]) > 10:
  print([1,2,3])
print("out of if")
```
