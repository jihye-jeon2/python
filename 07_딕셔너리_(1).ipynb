## 1. 딕셔너리의 개요
* **개념**: 리스트와 마찬가지로 여러 개의 데이터를 저장하는 자료형으로, **키(Key)**와 **값(Value)**의 쌍으로 이루어져 있습니다.
* **리스트와의 차이점**: 리스트는 인덱스(번호)를 사용하여 항목에 접근하지만, 딕셔너리는 사용자가 정의한 '키'를 사용하여 데이터(값)에 접근합니다.
* **생성 방법**: 중괄호 `{ }`를 사용하며, 각 항목은 `키:값` 형식으로 작성하고 쉼표(`,`)로 구분합니다.
* **키의 조건**: 딕셔너리 내에서 키는 중복될 수 없으며 유일해야 합니다. 만약 동일한 키가 여러 개 있다면 마지막 항목만 유효합니다.

```python
prices={'아메리카노':2000,'라떼':3000,'아메리카노':1900,'녹차':2500}
print(prices) #키가 중복이 될경우엔 마지막 항목인 아메리카노 1900만 남는다
```

```python
print(prices['라떼'])
```

```python
prices={}
prices['아메리카노']=2000
prices['라떼']=3000
prices['녹차']=2500
print(prices)
```

```python
#아메리카노 가격을 바꿈
prices['아메리카노']=2500
print(prices)
```

```python
#딕셔너리에서 키만 가지고올수잇음
for key in prices:
  print(key,end=' ')
```

```python
for key in prices:
  print(prices[key])
```

```python
for key,value in prices.items():
  print(key,value) #키와 value를 하나씩 가져옴
```

```python
for key in prices.keys():
  print(key)
```

## 2. 딕셔너리 다루기
데이터를 조회, 추가, 삭제하는 방법입니다.
* **값에 접근**: 딕셔너리 이름 뒤 대괄호 `[ ]` 안에 키 이름을 넣어 값을 가져옵니다.
* **항목 추가 및 변경**: 새로운 키에 값을 대입하면 항목이 추가되고, 기존에 있는 키에 값을 대입하면 해당 값이 수정됩니다.
* **항목 삭제 (del)**: 특정 키를 지정하여 해당 항목(키와 값의 쌍)을 삭제합니다.
* **모든 항목 삭제 (clear)**: 딕셔너리 내부의 모든 데이터를 한꺼번에 비웁니다.

```python
#항목삭제: pop() 함수이용하기 , 항목을삭제하고값을반환하는역할을한대,지정한 위치에 잇는 항목을 삭제, 삭제되는 항목의값을 반환할수잇
a= prices.pop('라떼')
print(a)  #~키 값을 삭제햇다 알기 위해 값을 반환
```

```python
# del 키워드, 딕셔너리의 키 항목을 삭제함
prices = {'아메리카노':2000,'라떼': 3000,'녹차':2500}
del prices['라떼']
print(prices)
```

```python
#clear(): 딕셔너리의 모든항목을 삭제함
prices.clear()
print(prices)   #{}
```

## 3. 딕셔너리 전용 함수
데이터를 효율적으로 관리하기 위한 기능들입니다.
* **키 목록 가져오기 (keys)**: 딕셔너리에 포함된 모든 키들을 모아서 보여줍니다.
* **값 목록 가져오기 (values)**: 딕셔너리에 포함된 모든 값들을 모아서 보여줍니다.
* **쌍으로 가져오기 (items)**: 키와 값의 쌍을 튜플 형태로 묶어서 반환하며, 반복문과 함께 자주 사용됩니다.
* **키 존재 확인 (in)**: 특정 키가 딕셔너리에 들어있는지 확인하여 참/거짓을 반환합니다.

```python
#국가이름 입력받고 국가의 수도를 알려주는 식
country_capital={'한국':'서울','일본':'도쿄','중국':'베이징'}
country =input('국가이름을 입력하세요: ')
if country in country_capital:
  print(country_capital[country])
else:
  print(f'{country}는 등록되지 않앗음')
```

```python
#음료 종류와 주문 수를 입력받아 주문 금액 계산하기
prices={'아메리카노':2000,'라떼':3000,'녹차':2500}
sum= 0
while True:
  s = input('음료 종류 또는 종료: ')
  if s=='종료':
    break
  elif s in prices:
    n=int(input(f'{s}주문수'))
    sum+=prices[s] *n
  else:
    print(f'{s}는 없는 음료입니다.')
print(f'총금액:{sum}')
```

```python
#혈액형을 입력 받는 동작을 '종료'가 입력될때까지 반복하여 각 혈액형의 빈도수 구하는 식
#결과 : A 1,B 2,O 1,AB 0
blood={'A':0,'B':0,'O':0,'AB':0}
while True:
  s=input('혈액형 입력 또는 종료: ')
  if s=='종료':
    break
  elif s in blood:
    blood[s]+=1
  else:
    print(f'{s}는 없는 혈액형')
for key,value in blood.items():
  print(key,value)
```

### 📊 프로그래밍 언어 선호도 조사
1. **데이터 수집**: 사용자로부터 사용 가능한 프로그래밍 언어를 여러 번 입력받습니다.
2. **카운팅 로직**:
    * 입력받은 언어가 이미 딕셔너리의 키로 존재하면, 해당 키의 값을 1 증가시킵니다.
    * 존재하지 않는 새로운 언어라면, 해당 언어를 키로 추가하고 값에 1을 저장합니다.
3. **결과 출력**: 각 언어별로 몇 번씩 입력되었는지 집계된 데이터를 보여줍니다.

```python
#선호하는 프로그램 조사

langs={'Python':0,'C':0,'JavaScript':0}
while True:
  s=input('선호하는 언어(Python,C,Java,JavaScript) 또는 종료: ')
  if s=='종료':
    print('종료합니다')
    break
  elif s in langs:
    langs[s]+=1
  else:
    print(f'{s}는 없는 언어')

for key,value in langs.items():
  print(key,value)
```

```python
#사용 가능한 프로그래밍 언어 조사
langs= {}

while True:
  s=input('사용 가능한 언어를 모두 입력하세요(쉼표,로 구분해서 입력)또는 종료: ')
  s=s.lower() # 입력받은걸 전부 소문자로 바꿔주기
  if s=='종료':
    print('종료합니다.')
    break
 # s = s.split(',')
  #print(s)

  for a in s.split(','):
    if a in s:
      langs[a]+=1
    else:
      langs[a]=1
      for key,value in langs.items():
        print(key,value)
```

### 🦁 영어 단어 퀴즈 (동물 이름)
1. **데이터 준비**: 한국어 동물 이름을 키로, 영어 단어를 값으로 하는 딕셔너리를 생성합니다.
2. **퀴즈 실행**: `items()` 함수를 사용하여 딕셔너리의 모든 항목을 순회합니다.
3. **사용자 입력**: 한국어 이름을 제시하고 사용자에게 영어 단어를 입력받습니다.
4. **정답 판별**: 사용자가 입력한 단어와 딕셔너리의 '값(영어 단어)'이 일치하는지 비교하여 정답 여부를 출력합니다.

```python
words={'사자':'lion','호랑이':'tiger','토끼':'rabbit'}
for key, value in words.items():
  answer = input(f'{key}의 영어이름은 ?: ')
  if answer ==  value:
    print('맞았습니다')
  else:
    print('틀렷습니다')
```
